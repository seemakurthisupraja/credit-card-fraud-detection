```python
duplicates = df[df.duplicated(keep=False)]

print("Total duplicate rows:", len(duplicates))
print("\nClass distribution among duplicates:")
print(duplicates["Class"].value_counts())
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
X = df.drop("Class", axis=1)
y = df["Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Training class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = X_train.copy()
X_test = X_test.copy()

X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
print("Training Time mean:", X_train["Time"].mean())
print("Training Time std:", X_train["Time"].std())

print("Training Amount mean:", X_train["Amount"].mean())
print("Training Amount std:", X_train["Amount"].std())
```

```python
class_counts = y_train.value_counts()

genuine_count = class_counts[0]
fraud_count = class_counts[1]

print("Genuine transactions:", genuine_count)
print("Fraud transactions:", fraud_count)
print("Genuine-to-Fraud ratio:", round(genuine_count / fraud_count, 2))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model_lr = LogisticRegression(
    class_weight="balanced",
    random_state=42,
    max_iter=1000
)
```

```python
model_lr.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
y_pred_lr = model_lr.predict(X_test)

print("Predictions generated successfully!")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_lr))
```

```python
cm_lr = confusion_matrix(y_test, y_pred_lr)

print(cm_lr)
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    xticklabels=["Genuine", "Fraud"],
    yticklabels=["Genuine", "Fraud"]
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
tn, fp, fn, tp = cm_lr.ravel()

print("True Negatives :", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives :", tp)
```

```python
precision = tp / (tp + fp)
recall = tp / (tp + fn)

print("Fraud Precision:", round(precision, 4))
print("Fraud Recall   :", round(recall, 4))
```

```python
y_pred_rf = model_rf.predict(X_test)

print("Random Forest predictions generated successfully!")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_rf))
```

```python
cm_rf = confusion_matrix(y_test, y_pred_rf)

print(cm_rf)
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    xticklabels=["Genuine", "Fraud"],
    yticklabels=["Genuine", "Fraud"]
)

plt.title("Random Forest - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_xgb,
    annot=True,
    fmt="d",
    xticklabels=["Genuine", "Fraud"],
    yticklabels=["Genuine", "Fraud"]
)

plt.title("XGBoost - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision_lr, recall_lr, _ = precision_recall_curve(
    y_test,
    y_prob_lr
)

precision_rf, recall_rf, _ = precision_recall_curve(
    y_test,
    y_prob_rf
)

precision_xgb, recall_xgb, _ = precision_recall_curve(
    y_test,
    y_prob_xgb
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall_lr,
    precision_lr,
    label="Logistic Regression"
)

plt.plot(
    recall_rf,
    precision_rf,
    label="Random Forest"
)

plt.plot(
    recall_xgb,
    precision_xgb,
    label="XGBoost"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid()

plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_prob_xgb
)

print("Number of thresholds:", len(thresholds))
```

```python
import numpy as np

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = np.argmax(f1_scores)

best_threshold = thresholds[best_index]
best_f1 = f1_scores[best_index]

print("Best threshold:", round(best_threshold, 4))
print("Best F1-score:", round(best_f1, 4))
```

```python
print(
    "Optimal threshold for F1:",
    round(best_threshold, 4)
)

print(
    "Maximum F1-score:",
    round(best_f1, 4)
)
```

```python
y_pred_xgb_tuned = (
    y_prob_xgb >= best_threshold
).astype(int)

print("Tuned predictions generated successfully!")
```

```python
print(classification_report(y_test, y_pred_xgb_tuned))
```

```python
cm_xgb_tuned = confusion_matrix(
    y_test,
    y_pred_xgb_tuned
)

print(cm_xgb_tuned)
```

```python
from sklearn.model_selection import train_test_split

X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

print("Final training features:", X_train_final.shape)
print("Validation features:", X_val.shape)
print("Final training fraud:", y_train_final.sum())
print("Validation fraud:", y_val.sum())
```

```python
scale_pos_weight_final = (
    y_train_final.value_counts()[0]
    / y_train_final.value_counts()[1]
)

print("Final scale_pos_weight:", scale_pos_weight_final)
```

```python
model_xgb_final = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=scale_pos_weight_final,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)
```

```python
model_xgb_final.fit(
    X_train_final,
    y_train_final
)

print("Final XGBoost model trained successfully!")
```

```python
y_prob_val = model_xgb_final.predict_proba(X_val)[:, 1]

print("Validation probabilities generated successfully!")
```

```python
precision_val, recall_val, thresholds_val = precision_recall_curve(
    y_val,
    y_prob_val
)

f1_scores_val = (
    2 * precision_val[:-1] * recall_val[:-1]
    / (precision_val[:-1] + recall_val[:-1] + 1e-10)
)

best_index_val = np.argmax(f1_scores_val)

best_threshold_val = thresholds_val[best_index_val]
best_f1_val = f1_scores_val[best_index_val]

print("Best validation threshold:", round(best_threshold_val, 4))
print("Best validation F1-score:", round(best_f1_val, 4))
```

```python
y_pred_val = (
    y_prob_val >= best_threshold_val
).astype(int)

print(classification_report(y_val, y_pred_val))
```

```python
y_prob_test_final = model_xgb_final.predict_proba(X_test)[:, 1]

y_pred_test_final = (
    y_prob_test_final >= best_threshold_val
).astype(int)

print("Final test predictions generated successfully!")
```

```python
print("FINAL TEST RESULTS")
print(classification_report(y_test, y_pred_test_final))
```

```python
cm_final = confusion_matrix(
    y_test,
    y_pred_test_final
)

print("Final Confusion Matrix:")
print(cm_final)
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_final,
    annot=True,
    fmt="d",
    xticklabels=["Genuine", "Fraud"],
    yticklabels=["Genuine", "Fraud"]
)

plt.title("Final XGBoost - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
import joblib
import os

os.makedirs("models", exist_ok=True)

model_package = {
    "model": model_xgb_final,
    "scaler": scaler,
    "threshold": best_threshold_val,
    "features": X_train.columns.tolist()
}

joblib.dump(
    model_package,
    "models/fraud_detection_model.pkl"
)

print("Model saved successfully!")
```

```python
loaded_package = joblib.load(
    "models/fraud_detection_model.pkl"
)

loaded_model = loaded_package["model"]
loaded_scaler = loaded_package["scaler"]
loaded_threshold = loaded_package["threshold"]
loaded_features = loaded_package["features"]

print("Model loaded successfully!")
print("Threshold:", loaded_threshold)
print("Number of features:", len(loaded_features))
```

```python
loaded_probabilities = loaded_model.predict_proba(
    X_test
)[:, 1]

loaded_predictions = (
    loaded_probabilities >= loaded_threshold
).astype(int)

print(
    "Predictions identical:",
    np.array_equal(
        loaded_predictions,
        y_pred_test_final
    )
)
```
